```python
dataset_locations = []

def label_extractor(location, included_labels, dataset_locations):
  # included_labels: The labels you want to keep
  for directory in ["train", "valid", "test"]:
    data_directory = f"{location}/{directory}"
    utils.filter_files_by_labels(data_directory, included_labels)

    label_path = f"{location}/{directory}/labels/"
    for target_label in included_labels:
      utils.replace_whole_numbers_in_files(label_path, str(target_label), "0")
  dataset_locations.append(location)
```
